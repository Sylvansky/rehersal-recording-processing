## Parse Song Versions from Filenames
Create a first-pass catalog of song instances and versions using title/date parsing.

```python
instances_df, summary_df = build_song_version_catalog(config["FILES_CSV"], exclude_hidden=True)
out_dir = config["ANALYZED_DIR"]
out_dir.mkdir(parents=True, exist_ok=True)

instances_path = out_dir / "song_instances.csv"
summary_path = out_dir / "song_versions_summary.csv"
instances_df.to_csv(instances_path, index=False)
summary_df.to_csv(summary_path, index=False)

print(f"Saved instances: {instances_path}")
print(f"Saved summary:   {summary_path}")
print()
print(f"Song groups: {len(summary_df)}")
summary_df.head(20)
```
